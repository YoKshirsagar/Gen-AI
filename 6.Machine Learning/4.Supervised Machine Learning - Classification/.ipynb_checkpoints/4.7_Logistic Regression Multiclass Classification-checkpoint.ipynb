```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Load Data

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris(as_frame = True)
iris
```

```python
df = iris.frame
df.head()
```

```python
df.target.unique()
```

```python
df.shape
```

```python
iris.target_names
```

```python
print(iris.DESCR)
```

```python
iris.keys()
```

### Data Visualization

```python
df['target_name'] = df['target'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
df.head()
```

```python
sns.scatterplot(df, x="sepal length (cm)", y="sepal width (cm)", hue="target_name")
```

```python
sns.scatterplot(df, x="petal length (cm)", y="petal width (cm)", hue="target_name")
```

```python
X = df[["sepal length (cm)", "sepal width (cm)", "petal length (cm)", "petal width (cm)"]]
y = df["target"]
X.head()
```

### Model Training

```python
from sklearn.linear_model import LogisticRegression

# Split dataset
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=5)

# Train logistic regression model 
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)
```

### Model Evaluation: Accuracy

```python
model.score(X_test, y_test)
```

### Model Evaluation: Precision, Recall, F1-Score

```python
y_pred = model.predict(X_test)
y_pred[:10]
```

```python
y_test[:10].tolist()
```

```python
from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)
```

### Model Evaluation: Confusion Matrix

```python
iris.target_names
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, xticklabels=iris.target_names, yticklabels=iris.target_names)   
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.title('Confusion Matrix')
plt.show()
```
