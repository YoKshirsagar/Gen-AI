```python
import pandas as pd
df = pd.read_csv("9.4_HandleNAValues-Part1_weather_data.csv",parse_dates=["day"])
df
```

```python
type(df.day)
```

```python
type(df.day[0])
```

```python
df.set_index('day',inplace=True)
```

```python
df
```

```python
# df.fillna(0,inplace=True)
df.fillna({
    'temperature': df.temperature.mean(),
    'windspeed':df.windspeed.mean(),
    'event': 'No Event'
}, inplace=True)
df
```

```python
df.ffill()           #Fill Null value to its previous row value
```

```python
df.bfill()           #Fill Null value to its next row value
```

```python
df.bfill(axis='columns')     #Fill Null value to its next column value
```

```python
df.bfill(limit=1)           #Fill Null value to its next row value for only next 1 Null Value
```

```python
df
```

```python
df.select_dtypes(include='number').interpolate()    #Fill Null value to mid of next row value + previous row value
```

```python
df
```

```python
df.dropna(how="any")          #Drop row having atleast 1 NA Value
```

```python
df.dropna(how="all")   #Drop row having All Value as NA
```

```python
df.dropna(thresh=2)    #Drop row having atleast 2 NA Values
```
