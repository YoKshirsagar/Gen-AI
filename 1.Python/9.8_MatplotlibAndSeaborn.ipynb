```python
import pandas as pd
```

```python
df_sales=pd.read_excel("9.8_MatplotlibAndSeaborn_linechart.xlsx")
df_sales
```

```python
from matplotlib import pyplot as plt 
```

```python
plt.figure(figsize=(14,4))
plt.plot(df_sales["Quarter"],df_sales["Fridge"])
plt.plot(df_sales["Quarter"],df_sales["Dishwasher"])
plt.plot(df_sales["Quarter"],df_sales["Washing Machine"])
plt.show()
```

```python
plt.figure(figsize=(14,4))
plt.plot(df_sales["Quarter"],df_sales["Fridge"],color="red",label="Fridge")
plt.plot(df_sales["Quarter"],df_sales["Dishwasher"],color="black",label="Dishwasher")
plt.plot(df_sales["Quarter"],df_sales["Washing Machine"],color="blue",label="Washing Machine")
plt.title("Product Sales")
plt.ylabel("Revenue ($)")
plt.xlabel("Financial Quarter")
plt.legend()
plt.show()
```

```python
total_sales =df_sales[["Fridge","Dishwasher","Washing Machine"]].sum()
total_sales
```

```python
total_sales.index
```

```python
plt.pie(total_sales,labels=total_sales.index)
plt.show()
```

```python
plt.pie(total_sales,labels=total_sales.index,autopct="%1.1f%%")
plt.show()
```

```python
plt.pie(total_sales,labels=total_sales.index,autopct="%1.1f%%",explode=(0.1,0,0))
plt.show()
```

```python
plt.pie(total_sales,labels=total_sales.index,autopct="%1.1f%%",explode=(0.1,0,0),shadow=True)
plt.show()
```

```python
plt.pie(total_sales,labels=total_sales.index,autopct="%1.1f%%",explode=(0.1,0,0),shadow=True,startangle=140)
plt.show()
```
