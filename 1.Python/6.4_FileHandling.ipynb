```python
player_scores={}
with open("6.4_FileHandlingReadFile.csv",'r') as f:
    for line in f:
        player , _ , score = line.split(',')
        score = int(score.strip()) # int('56\n'.strip())
        if player in player_scores:
            player_scores[player].append(score)
        else:
            player_scores[player] = [score]
            
        
```

```python
print(player_scores)
```

```python
for player,score_list in player_scores.items():
    min_score = min(score_list)
    max_score = max(score_list)
    avg_score = sum(score_list)/len(score_list)
    print(f"{player} ==> Max:{max_score} | Min:{min_score} | Avg:{avg_score}")
    
```
